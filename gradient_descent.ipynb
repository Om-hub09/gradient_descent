```python
from sklearn.datasets import make_regression
import numpy as np
```

```python
x,y=make_regression(n_samples=4,n_features=1,n_informative=1,n_targets=1,noise=80,random_state=13)
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
plt.scatter(x,y)
```

```python
from sklearn.linear_model import LinearRegression
reg=LinearRegression()
reg.fit(x,y)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
plt.scatter(x,y)
plt.plot(x,reg.predict(x),color='red')
```

```python
##lets apply gradient descent when slope(m) =78.35063668 and intercept(b)=0(lets say as starting value)
y_pred=((78.35063668*x)+0).reshape(4)
```

```python
plt.scatter(x,y)
plt.plot(x,reg.predict(x),color='red',label="OLS")
plt.plot(x,y_pred,color='green',label="b=0")
plt.legend()
plt.show()
```

```python
## 1st iteration
m=78.35063668
b=0

loss_slope= -2* np.sum(y-x.ravel() -b)
print(loss_slope)
```

```python
#lets take learning rate as 0.1
lr=0.1
step_size=loss_slope*lr
print(step_size)
```

```python
# calculating the step size
b=b-step_size
print(b)
```

```python
y_pred1=((78.35063668*x)+b).reshape(4)
plt.scatter(x,y)
plt.plot(x,reg.predict(x),color='black',label='ols')
plt.plot(x,y_pred1,color='red',label='b={}'.format(b))
plt.plot(x,y_pred,color='green',label='b=0')
plt.legend()
plt.show()
```

```python
#2nd iteration
loss_slope2=-2*np.sum(y-x.ravel()-b)
print(loss_slope2)
```

```python
step_size=loss_slope2*lr
step_size
```

```python
b=b-step_size
b
```

```python
y_pred2=((78.35063668*x)+b).reshape(4)

plt.scatter(x,y)
plt.plot(x,reg.predict(x),color='red',label="OLs")
plt.plot(x,y_pred2,color='blue',label='b={}'.format(b))
plt.plot(x,y_pred1,color='green',label='b={}'.format(b))
plt.plot(x,y_pred,color='yellow',label='b=0')
plt.legend()
plt.show()
```

```python
#3rd Iteration
loss_slope3=-2*np.sum(y-x.ravel()-b)
loss_slope3
```

```python
step_size=loss_slope3*lr
step_size
```

```python
b=b-step_size
b
```

```python
y_pred3=((78.35063668*x)+b).reshape(4)

plt.figure(figsize=(15,15))
plt.scatter(x,y)
plt.plot(x,reg.predict(x),color='red',label="OLs")
plt.plot(x,y_pred3,color='orange',label='b={}'.format(b))
plt.plot(x,y_pred2,color='blue',label='b={}'.format(b))
plt.plot(x,y_pred1,color='green',label='b={}'.format(b))
plt.plot(x,y_pred,color='yellow',label='b=0')
plt.legend()
plt.show()
```

```python
b=-100
m=78.35
lr=0.1
epochs=10

for i in range(epochs):
    loss_slope= -2 * np.sum(y-m*x.ravel()-b)
    b=b-(lr*loss_slope)
    y_pred=m*x.ravel()-b
    plt.plot(x,y_pred)
plt.scatter(x,y)

```
